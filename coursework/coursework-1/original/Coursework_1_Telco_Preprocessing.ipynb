# Coursework 1 — Telco Customer Churn Data Preparation
GitHub portfolio edition based on the uploaded coursework notebook. The original execution recorded an initial shape of **(7043, 21)** and exported processed data, summary statistics and visualisations.


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
df = df.drop_duplicates().drop(columns=['customerID'])
df = pd.get_dummies(df, drop_first=True)
scaler = StandardScaler()
numeric = ['tenure', 'MonthlyCharges', 'TotalCharges']
df[numeric] = scaler.fit_transform(df[numeric])
df.to_csv('telco_processed.csv', index=False)
```

## Recorded original output
- Dataset shape: 7,043 rows × 21 columns
- `TotalCharges` converted to numeric and median-imputed
- `customerID` removed
- categorical features encoded
- key numeric features standardised
- processed dataset + visual outputs exported
